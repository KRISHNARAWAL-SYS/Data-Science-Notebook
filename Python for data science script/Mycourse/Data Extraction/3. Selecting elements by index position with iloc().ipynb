# 3. Selecting a range of data with a slice

```python
# slice colm labels age/club
# players =['L.Messi' , 'CR7']
players = [0,1]
df.iloc[players , 1:7]
```

```python
# slice indec labels (top1 to top10 player)
# columns = ['age','dob','height_cm','weight_kg']

columns = [1,2,3,4]

# get top1 and top10 player name
df.iloc[0:10 , columns]
```

# 4. Selectinf with conditions

iloc cannot accept a boolean series , but olny a boolean list . we have to use the  list() function to convert a series into a boolean list

```python
# one condition select player with height above 180cm
# colimns = ['age' , 'dob' ,'height_cm' , 'weight_kg']

columns = [1,2,3,4]

list(df['height_cm']>180)

df.iloc[list(df['height_cm']>180) , columns]
```

```python
# Mulitple condition : select player with height above 180cm and argentina

list((df['height_cm']>180) & ( df['nationality'] == "Argentina"))

df.iloc[list((df['height_cm']>180) & ( df['nationality'] == "Argentina")) , columns]
```
