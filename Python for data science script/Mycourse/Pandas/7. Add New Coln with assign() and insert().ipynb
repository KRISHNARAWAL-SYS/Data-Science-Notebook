# Adding a new column

## assign()

#### when to use assign?
####     >Add multiple colns in a single line of code
####     >when you need to overwrite the value of an existing columns
####     > It returns a new ibject (a copy) with all the original columns in addition to the          new ones

```python
import numpy as np
```

```python
# Create random numbers for our new score columns
score1 = np.random.randint(1,100, size=1000)
score2 = np.random.randint(1,100, size=1000)
```

```python
# Create a series using the random numbers
series1 = pd.Series(score1 , index=np.arange(0,1000))
series2 = pd.Series(score2 , index=np.arange(0,1000))
```

```python
series1
```

```python
# using assign() to add multiple columns
df_exams=df_exams.assign(score1=series1 , score2=series2)
# assigning this code to df_exams is called overwriting update the data frame
```

```python
df_exams
```

## insert()
##### inserts a new column at a specific position or index

```python
# using index() to add a column at a specific postion 
df_exams.insert(1,'test',series1)
```
