# 1 Lambda Function

```python
# basic function
def sum_values(a,b):
    x = a+b
    return x
```

```python
sum_values(2,3)
```

```python
# Lambda function (One liner)
sum_values_lambda = lambda a,b: a+b
```

```python
sum_values_lambda(10,23)
# A lambda function is a temporary function which we might use only once 
```

# 2 Apply + Lambda FUcntion

```python
# use a lambda function to convert height_cm series to meters
df['height_cm'].apply(lambda x : x/100)

#alternative
df['height_cm']/100
```

```python
# use L.FUnction to convert "long_name" colm to uppercase
df['long_name'].apply(lambda x : x.upper())

# alternative with str attribute
df['long_name'].str.upper()
```

```python
# changing the datatype of "dob"
df['dob'] = df['dob'].astype('datetime64[ns]')
df.dtypes
```

```python
df['dob'] = df['dob'].astype('datetime64[ns]')
# use L. Function to get the year of "dob" series
df['dob'].apply(lambda x : x.year)

# alternative with dt attribute
df['dob'].dt.year
```

```python
# apply lambda function to dataframe in order to calculate BMI
df.apply(lambda x : x['weight_kg'] / ((x['height_cm']/100)**2) , axis=1)
```
