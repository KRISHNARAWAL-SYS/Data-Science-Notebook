```python
import pandas as pd
```

```python
# reading the csv file
df_exams = pd.read_csv(r"C:\Users\krishna rawal\dataset.csv")
df_exams.head()
```

```python
import numpy as np
```

```python
# Create random numbers for our new score columns
score1 = np.random.randint(1,100, size=1000)
score2 = np.random.randint(1,100, size=1000)
```

```python
# Create a series using the random numbers
series1 = pd.Series(score1 , index=np.arange(0,1000))
series2 = pd.Series(score2 , index=np.arange(0,1000))
```

```python
# USING ASSIGN() TO ADD MULTIPLE COLUMNS
df_exams = df_exams.assign(score1 = series1 ,score2 = series2)
```

```python
# printing dataFrame
df_exams
```

# 1 Sort a DataFrame

```python
# Sort by one column
df_exams.sort_values(by='score1')
```

```python
# Sort descending by one column
df_exams.sort_values(by='score1')
df_exams.sort_values('score1')   #same
```

```python
# Sort descending by multiple columns
df_exams.sort_values(by='score1' , ascending=False)
```

```python
# Sort descending by multiple columns
df_exams.sort_values(['score1','score2'], ascending=False)
```

```python
# Sort descending by multiple columns and update the dataframe
df_exams.sort_values(['score1','score2'], ascending=False,inplace=True)
# inplace overwrite or update the DataFrame
df_exams
```

```python
# sort descending with a key function
df_exams.sort_values('color' , ascending=True , key=lambda col:col.str.lower())
# note : you can sort the column as it is now , but we'll use the "key" argument to ignore case of text
```
