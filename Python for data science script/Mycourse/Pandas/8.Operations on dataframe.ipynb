```python
import pandas as pd
```

```python
# reading the csv file
df_exams = pd.read_csv(r"C:\Users\krishna rawal\dataset.csv")
df_exams.head()
```

```python
# FOR OPERATIONS I HAVE TO ADD COLUMNS
```

```python
import numpy as np
```

```python
# Create random numbers for our new score columns
score1 = np.random.randint(1,100, size=1000)
score2 = np.random.randint(1,100, size=1000)
```

```python
# Create a series using the random numbers
series1 = pd.Series(score1 , index=np.arange(0,1000))
series2 = pd.Series(score2 , index=np.arange(0,1000))
```

```python
# USING ASSIGN() TO ADD MULTIPLE COLUMNS
df_exams = df_exams.assign(score1 = series1 ,score2 = series2)
```

```python
# printing dataFrame
df_exams
```

# 1 Math operation

### 1.1 Operations on Columns

```python
# Select a column and calculate total sum
df_exams['score2'].sum()
```

```python
# count, mean, std, max and min
df_exams['score1'].count()
df_exams['score1'].mean()
df_exams['score1'].std()      # standard deviation
df_exams['score1'].max()
df_exams['score1'].min()
```

```python
# Easier Calculations with .describe()
df_exams.describe()
```

### 1.2 Operations in rows 

```python
# Calculating the sum in a row
df_exams['score1'] + df_exams['score2']
```

```python
# calculating the average score and assigning the result to a new column 
df_exams['Avg_Score'] = (df_exams['score1'] + df_exams['score2'])/2
```
