#  1. drop_duplicate()

```python
# drop duplicates in 2 or more columns
df_laptops.drop_duplicates(['Company'])
```

```python
# to verify
df_laptops.drop_duplicates(['Company'])[['Company','Price_euros']]
```

```python
df_laptops.drop_duplicates(['Company'])[['Company','Price_euros']].value_counts('Company')
```

```python
# sort dataframe ascending by company and price
# cheap first and most expensive last 
df_laptops=df_laptops.sort_values(['Company','Price_euros'])
df_laptops
```

```python
# cheapest : keep ='first'
df_laptops.drop_duplicates(['Company'], keep = 'first')[['Company','Price_euros']]
```

```python
# expensive : keep = 'last'
df_laptops.drop_duplicates(['Company'], keep = 'last' , inplace = True , ignore_index=True)
# inplace = true ---> all the changes made to the original dataframe are to be saved 
df_laptops
```

```python
# Arguments

# inplace: drop duplicates in place or to return a copy.
# ignore_index = True: the resulting axis will be labelled 0,1,...
df_laptops[['Company','Price_euros']]
```

Task : find the biggest and smallest screen size in laptops of each company using the sort_values() and duplicated() methods (keep "first" , "last" and False)

```python
df_laptops = pd.read_csv(r"C:\Users\krishna rawal\OneDrive\Desktop\laptop_price.csv" , encoding='latin1')
```

```python
# sort dataframe ascnding by company and inches
# Smallest first and biggest last
df_laptops = df_laptops.sort_values(['Company' , 'Inches'])
```

```python
# Smallest : keep = 'First'
df_laptops.drop_duplicates(['Company'], keep = 'first')[['Company','Inches']]
```

```python
# Biggest : keep = 'last'
df_laptops.drop_duplicates(['Company'] , keep = 'last')[['Company' , 'Inches']]
```
