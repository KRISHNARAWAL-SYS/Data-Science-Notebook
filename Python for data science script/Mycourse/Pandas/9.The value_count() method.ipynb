# 1. The value count

```python
# counting gender elements

# len function
len(df_exams['gender'])

# .count() method
df_exams['gender'].count()
```

```python
# counting gender elements by category
df_exams['gender'].value_counts()
```

```python
# return the relative frequency(divide all values by the sum of values)
df_exams['gender'].value_counts(normalize=True)*100 # (in percentage)
```

```python
# counting " color" elements by category
df_exams['color'].value_counts()
```

```python
#  return the relative frequency and round to 2 decimals
df_exams['color'].value_counts() 
```

```python
# return the relative frequency and round to 2 decimals
df_exams['color'].value_counts(normalize=True).round(2)
```
