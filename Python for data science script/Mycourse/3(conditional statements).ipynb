# Conditional Statements

### If Statements

```python
if <condition>:
    <code>
elif <condition>:
    <code>
...
else:
    <code>
```

```python
age = int(input("enter the age :"))
if age>=18:
    print('Can Vote')
else:
    print('Cannot Vote')
    
```

### For Loop

```python
for <variables> in <iterable>:
    <code>
```

```python
countries = ['united States','India','China','Brazil']
```

```python
for place in  countries:
    print(place)
```

```python
for i , country in enumerate(countries):
    print(i)
    print(country)
```

```python
my_data = {'Name':'KRishnaRawal' , 'Age' :'20'}
my_data
```

```python
for key in my_data:
    print(key)
```

```python
for value in my_data:    # NO difference
    print(value)
```

```python
for value in my_data.items():   # Quite of a progress
    print(value)
```

```python
for key , value in my_data.items():
    print(key)
    print(value)
```
