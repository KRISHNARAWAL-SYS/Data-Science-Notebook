# 2 THe Split-Apply-Combine Strategy

```python
# split the data into separate groups
passenger_filter = df_cars['Vehicle_type'] == 'Passenger'
car_filter = df_cars['Vehicle_type'] == 'Car'
# df_cars['Vehicle_type'].value_counts()
```

```python
# apply an operation
passenger_mean = df_cars[passenger_filter]['Sales_in_thousands'].mean()
car_mean       = df_cars[car_filter]['Sales_in_thousands'].mean()
```

```python
# combine results
pd.DataFrame({'Vehicle_type' : ['Car' , 'Passenger'],
             'Sales_in_thousands' : [car_mean , passenger_mean ]})
```
