```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv(r"C:\Users\krishna rawal\OneDrive\Desktop\players_20.csv")
# set index
df.set_index('short_name', inplace = True)  
# Select column
df= df[['long_name','age','dob','height_cm','weight_kg','nationality','club']]
```

```python
# show dataframe
df
```

# 1 Drop Row


```python
# drop one row
# axis parameter
df.drop('L. Messi' , axis=0)  # 0 represents rows
# index parameter
df.drop( index= ['L. Messi'])   # same result u can choose any
```

```python
# drop two or more rows and update data (inplace=True)
df.drop(['L. Messi','Cristiano Ronaldo'] , axis = 0 , inplace = True)
df
```

# Drop Columns

```python
# drop one column
# axis parameter
df.drop('long_name' , axis =1)

# columns parameter
df.drop(columns = ['long_name'])
```

```python
# drop colm by position/index (drop last column)


df.drop(df.columns[[-1]] , axis=1)
```

```python
# drop two or more columns and update data (inplace=True) 
df.drop(['long_name','dob'] , axis =1 , inplace=True)
```
