# Creating a dataframe from an Array

### Option1

```python
#creating an array
data = np.array([[1,4],[2,4],[8,4]])
```

```python
#creating a dataframe
df=pd.DataFrame(data)
```

```python
# Showing the Dataframe
df
```

### Option2

```python
#creating an array
data = np.array([[11,4],[2,4],[8,9]])
```

```python
#creating a dataframe
df = pd.DataFrame(data , index=['row1','row2','row3'],
                  columns=['col1','col2'])
```

```python
# Showing the Dataframe
df
```

# Creating a DataFrame from a dictionary

```python
# lists used for this example
States = ['goa' , 'rajhasthan' , 'odissa' , 'sikkim']
population = [23435545 , 5522343535, 532344234, 3423434]
```

```python
#Storing lists within a dictionary
dict_states = {'States':States , 'population':population}
```

```python
#Creating the dataframe
df_population=pd.DataFrame(dict_states)
```

```python
# Showing the dataframe
df_population
```
